```python
import pandas as pd

# Yêu cầu 1: Tạo DataFrame chứa thông tin của 10 sinh viên
data = {
    'Họ và tên': [
        'Nguyễn Văn An', 'Trần Thị Bình', 'Lê Văn Cường', 'Phạm Thị Dung', 'Hoàng Văn Em',
        'Vũ Thị Phượng', 'Đỗ Văn Giang', 'Bùi Thị Hoa', 'Ngô Văn Hùng', 'Dương Thị Lan'
    ],
    'Chuyên cần': [8.5, 9.0, 7.0, 10.0, 6.5, 8.0, 9.5, 7.5, 8.5, 9.0],
    'Giữa kỳ': [7.0, 8.5, 6.0, 9.0, 5.5, 7.5, 8.0, 6.5, 7.0, 8.5],
    'Cuối kỳ': [7.5, 8.0, 6.5, 9.5, 6.0, 7.0, 8.5, 7.0, 8.0, 9.0]
}

df = pd.DataFrame(data)

# Yêu cầu 2: Tính điểm tổng kết (20% Chuyên cần + 30% Giữa kỳ + 50% Cuối kỳ)
df['Tong_ket'] = (0.2 * df['Chuyên cần']) + (0.3 * df['Giữa kỳ']) + (0.5 * df['Cuối kỳ'])
df['Tong_ket'] = df['Tong_ket'].round(2)

# Yêu cầu 3: Xếp loại sinh viên
bins = [-float('inf'), 5.0, 7.0, 8.5, float('inf')]
labels = ['Yếu', 'Trung bình', 'Khá', 'Giỏi']
df['Xep_loai'] = pd.cut(df['Tong_ket'], bins=bins, labels=labels, right=False)

# Yêu cầu 4: Thống kê kết quả lớp học
diem_tb_lop = df['Tong_ket'].mean()
max_diem = df['Tong_ket'].max()
min_diem = df['Tong_ket'].min()

sv_cao_nhất = df[df['Tong_ket'] == max_diem]['Họ và tên'].tolist()
sv_thap_nhất = df[df['Tong_ket'] == min_diem]['Họ và tên'].tolist()
so_sv_dat = len(df[df['Tong_ket'] >= 5.0])

print(" BẢNG KẾT QUẢ: ")
print(df)
print("\n THỐNG KÊ: ")
print(f"Điểm tổng kết trung bình của lớp: {diem_tb_lop:.2f}")
print(f"Sinh viên có điểm tổng kết cao nhất: {', '.join(sv_cao_nhất)} ({max_diem})")
print(f"Sinh viên có điểm tổng kết thấp nhất: {', '.join(sv_thap_nhất)} ({min_diem})")
print(f"Số sinh viên đạt (điểm tổng kết >= 5): {so_sv_dat}/10")
import matplotlib.pyplot as plt
import pandas as pd

# Cấu hình font chữ hỗ trợ tiếng Việt (nếu cần thiết trên hệ thống của bạn)
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.unicode_minus'] = False

# 1. Tạo dữ liệu và tính điểm tổng kết như các bước trước
data = {
    'Họ và tên': [
        'Nguyễn Văn An', 'Trần Thị Bình', 'Lê Văn Cường', 'Phạm Thị Dung', 'Hoàng Văn Em',
        'Vũ Thị Phượng', 'Đỗ Văn Giang', 'Bùi Thị Hoa', 'Ngô Văn Hùng', 'Dương Thị Lan'
    ],
    'Chuyên cần': [8.5, 9.0, 7.0, 10.0, 6.5, 8.0, 9.5, 7.5, 8.5, 9.0],
    'Giữa kỳ': [7.0, 8.5, 6.0, 9.0, 5.5, 7.5, 8.0, 6.5, 7.0, 8.5],
    'Cuối kỳ': [7.5, 8.0, 6.5, 9.5, 6.0, 7.0, 8.5, 7.0, 8.0, 9.0]
}

df = pd.DataFrame(data)
df['Tong_ket'] = (0.2 * df['Chuyên cần']) + (0.3 * df['Giữa kỳ']) + (0.5 * df['Cuối kỳ'])
df['Tong_ket'] = df['Tong_ket'].round(2)

# 2. Vẽ biểu đồ cột (Bar Chart)
plt.figure(figsize=(12, 6))
bars = plt.bar(df['Họ và tên'], df['Tong_ket'], color='skyblue', edgecolor='navy', width=0.6)

# Thêm tiêu đề và nhãn các trục
plt.title('BIỂU ĐỒ THỂ HIỆN ĐIỂM TỔNG KẾT CỦA 10 SINH VIÊN', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Họ và tên sinh viên', fontsize=12, labelpad=10)
plt.ylabel('Điểm tổng kết', fontsize=12, labelpad=10)

# Xoay tên sinh viên góc 30 độ để hiển thị rõ hơn, không bị đè chữ
plt.xticks(rotation=30, ha='right', fontsize=10)
plt.ylim(0, 10)  # Thang điểm từ 0 đến 10

# Hiển thị giá trị điểm số trực tiếp trên mỗi cột
for bar in bars:
    height = bar.get_height()
    plt.annotate(f'{height}',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),  # Khoảng cách 3 points lên phía trên
                textcoords="offset points",
                ha='center', va='bottom', fontsize=10, fontweight='bold')

# Thêm lưới ngang mờ giúp dễ quan sát giá trị
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Tự động căn chỉnh lề tránh bị cắt chữ
plt.tight_layout()

# Hiển thị biểu đồ
plt.show()
```
